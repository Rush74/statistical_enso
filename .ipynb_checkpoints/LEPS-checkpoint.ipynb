```python
import xarray as xr
import geopandas as gpd
import regionmask
import geopandas as gpd
from matplotlib.patches import Rectangle

base = "/home/563/dy2988/.jupyter-root/home/statistical_enso/data/BoM_scores/"

ds_0lead = xr.open_dataset(base + "rain.skill.leps.seasonal.tercile.ini01.lt0.09.glb.nc")
ds_1lead = xr.open_dataset(base + "rain.skill.leps.seasonal.tercile.ini01.lt1.08.glb.nc")
ds_2lead = xr.open_dataset(base + "rain.skill.leps.seasonal.tercile.ini01.lt2.07.glb.nc")
ds_3lead = xr.open_dataset(base + "rain.skill.leps.seasonal.tercile.ini01.lt3.06.glb.nc")
```

```python
mdb = gpd.read_file("~/data/mdb_boundary.json")

mdb_mask = regionmask.mask_geopandas(
    mdb,
    ds_0lead.lon,
    ds_0lead.lat
).notnull()

ds = [ds_0lead, ds_1lead, ds_2lead, ds_3lead]
mdb_leps = []

for i in range(4):
    leps = ds[i].where(mdb_mask)
    leps_mean = leps.mean(dim=['lat', 'lon'])
    mdb_leps.append(leps_mean)
    print(f'{i}-month lead forecast 1981-2018 SON - ACCESS-S2 LEPS: '
          f'{mdb_leps[i].leps.item():.2f}%')
print()

print('0 month-lead forecast 1981-2018 SON - LOOCV -  RONI LEPS: 10.55% , ONI LEPS: 10.37% ')
print('1 month-lead forecast 1981-2018 SON - LOOCV -  RONI LEPS: 7.89%, ONI LEPS: 6.80% ')
print('2 month-lead forecast 1981-2018 SON - LOOCV -  RONI LEPS: 1.75%, ONI LEPS: 1.17% ')
# print('3 month-lead forecast 2021 SON - sliding 25yrs window - RONI LEPS: 16.18%, ONI LEPS: 11.87% ')


print()

print('0 month-lead forecast 1981-2018 SON - TimeSplitSeries -  RONI LEPS: 15.29%, ONI LEPS: 17.01% ')
print('1 month-lead forecast 1981-2018 SON - TimeSplitSeries -  RONI LEPS: 13.00%, ONI LEPS: 12.37% ')
print('2 month-lead forecast 1981-2018 SON - TimeSplitSeries -  RONI LEPS: 4.44%, ONI LEPS: 4.56% ')
# print('3 month-lead forecast 2021 SON - sliding 25yrs window - RONI LEPS: 16.18%, ONI LEPS: 11.87% ')

print()

print('0 month-lead forecast 1981-2018 SON - Sliding30yrsWindow - RONI LEPS: 8.70%, ONI LEPS: 8.27% ')
print('1 month-lead forecast 1981-2018 SON - Sliding30yrsWindow - RONI LEPS: 6.14%, ONI LEPS: 4.28% ')
print('2 month-lead forecast 1981-2018 SON - Sliding30yrsWindow - RONI LEPS: -0.38%, ONI LEPS: -1.08% ')
# print('3 month-lead forecast 2021 SON - base 1981-2018 - RONI LEPS: 6.78%, ONI LEPS: 4.50% ')
```
