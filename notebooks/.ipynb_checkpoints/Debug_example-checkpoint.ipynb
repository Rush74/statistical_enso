Some notes on debugging in VSCode.  First step
1) Run through the tutorial here : https://21centuryweather.github.io/21st-Century-Weather-Software-Wiki/vscode/debugging-python.html
2) Get VSCode running on your Mac an open the `statistical_enso` directory
3) Load `scripts/debug_LDA.py` in the File Explorer

Now, make sure to select your local environment that supports `sklearn`, `numpy`, `pandas` etc. By default, VSCode may use the base python interpreter shipped with your Mac OS (which is pretty limited) or the `base` environment in your `conda` installation. To change this, first click on the python environment button on the bottom right corner, and then select your desired environment from the pull-down menu.

<img src='images/Select_pyenv.png'>

Then, as in the tutorial, use the left mouse button to set a 'Breakpoint' as designated by a red dot at line 249, which corresponds to the function call `combo_method_SKL`

<img src='images/Breakpoint.png'>

Then load the VSCode debugger mode as described in the tutorial

<img src='images/Debug.png'>

Make sure to select 'Debug the currently active Python file'

<img src='images/Debug_active.png'>

If you've done all that correctly, the Debugger should start running and stop at your breakpoint.

<img src='images/Debug_window.png'>


Then hit 'Step Into' using the Control Buttons (or the function key short-cuts) until you have entered the fuction `combo_method_SKL`. Remember you can activate the `Debug Console` and type the names of variables you want to track the values of. You can also use the Variable viewer in the left-hand panel.

<img src='images/Debug_console.png'>
